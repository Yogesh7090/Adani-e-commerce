```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Sample data
# texts = ["The service was excellent", "I am very unhappy", "It was okay"]
texts = ["It was the best of times.", "t was the worst of times.", "I get very frustrated with this app as it doesnt save your progress when you read when you switch from different devices. Very inconvenient when you are typically on a computer but have to switch to your phone because you arent home. Not to mention, on mobile devices, you cannot click on the the images to make them larger, which really takes away from the text book because they are usually important charts with data and informations. Check out the graph below well i cant so."]

labels = ["positive", "negative", "neutral"]

# Vectorize text
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2, random_state=42)

# Train model
model = LogisticRegression()
model.fit(X_train, y_train)

# Evaluate
predictions = model.predict(X_test)
print(classification_report(y_test, predictions))
```
